# Data steward profile
This notebook can be downloaded as **{nb-download}`2026-04-09.ipynb`**

This semester (spring 2026), I started a [CAS in Data Stewardship](https://www.unil.ch/news/fr/1765966357311).


One of the most striking things is realizing how broad a field data stewardship is. I don't know anyone who is a data steward, but I know a lot of people who do data stewardship and/or already have many skills in that area. It is very unlikely that one person can have all the skills and knowledge required to take on all the responsibilities of data stewardship. However, data stewardship can be seen as a collaborative effort, in which people with different skills and knowledge work together to ensure that data is properly managed and used.


It is therefore important to identify the skills and knowledge that are most relevant to the specific context in which you, as a data steward, will be working. Based on the Data steward skills and abilities grid {footcite}`SwissDS-ENV2023-mf` and the Definition and profile of data professions {footcite}`SwissDS-ENV2023-dd`, you can quickly identify the skills you already have and those you need to develop further. You might not be able to master all of them, but having a good understanding of them will help you collaborate better with other data stewards!
This script helps you visualize your data steward profile based on your relevant skills, knowledge, and experience. Each chart compares my profile with one of the data roles.

````python
import pandas as pd
import plotly.graph_objects as go
from IPython.display import display, HTML

# Categories
categories = [
    'Plan & design',
    'Collect & capture',
    'Collaborate & analyse',
    'Manage, store & secure',
    'Archive & share',
    'Promote for reuse'
]

# Data
df = pd.DataFrame({
    'Data Manager': [4, 3, 3, 4, 4, 4],
    'Data Librarian': [4, 2, 1, 3, 4, 4],
    'Data Scientist': [0, 5, 5, 0, 0, 0],
    'Data Curator': [0, 4, 2, 4, 2, 0],
    'Data Archivist': [0, 0, 0, 4, 5, 4],
    'Me': [1, 4, 5, 4, 2, 1]
}, index=categories)

# Close the loop for radar chart, and wrap the long labels
labels = [c.replace(' & ', '<br>& ').replace(' for ', '<br>for ') for c in categories]
labels_closed = labels + [labels[0]]

# Colors (the charts stay dark in both site themes)
ME_COLOR = '#199e70'
ROLE_COLOR = '#d95926'
PANEL = '#0e1411'
INK = '#e4ebe6'
GRID = '#2c3d33'


def add_profile(fig, name, color, dash, symbol, fill):
    values = list(df[name]) + [df[name].iloc[0]]
    fig.add_trace(go.Scatterpolar(
        r=values,
        theta=labels_closed,
        name=name,
        mode='lines+markers',
        fill='toself',
        fillcolor=fill,
        line=dict(color=color, width=2, dash=dash),
        marker=dict(color=color, size=8, symbol=symbol),
        hovertemplate='%{theta}: %{r}<extra>' + name + '</extra>'
    ))


def role_chart(role):
    fig = go.Figure()
    add_profile(fig, role, ROLE_COLOR, 'dash', 'diamond', 'rgba(217, 89, 38, 0.10)')
    add_profile(fig, 'Me', ME_COLOR, 'solid', 'circle', 'rgba(25, 158, 112, 0.30)')
    fig.update_layout(
        title=dict(text=role, x=0.5, font=dict(size=16)),
        polar=dict(
            bgcolor=PANEL,
            radialaxis=dict(range=[0, 5], dtick=1, gridcolor=GRID, linecolor=GRID,
                            tickfont=dict(size=10, color=INK)),
            angularaxis=dict(rotation=90, direction='clockwise', gridcolor=GRID,
                             linecolor=GRID, tickfont=dict(size=10))
        ),
        legend=dict(orientation='h', x=0.5, xanchor='center', y=-0.12),
        margin=dict(l=70, r=70, t=60, b=40),
        height=420,
        font=dict(color=INK),
        paper_bgcolor=PANEL
    )
    return fig


# One small chart per role, in a grid that wraps on narrow screens
roles = [role for role in df.columns if role != 'Me']
charts = [
    role_chart(role).to_html(
        full_html=False,
        include_plotlyjs='cdn' if i == 0 else False,
        config=dict(responsive=True, displayModeBar=False)
    )
    for i, role in enumerate(roles)
]
display(HTML('<div class="vf-radar-grid">' + ''.join(charts) + '</div>'))
````

Given my engineering background, my current data steward profile looks like a combination of a data curator and a data scientist, with a strong focus on data collection and analysis. I lack background and practical experience in data governance, which I hope to acquire during this CAS.
